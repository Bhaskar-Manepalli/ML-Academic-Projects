```python
#Apply KNN algorithm for classification.

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

iris=pd.read_csv('/content/drive/MyDrive/Iris (2).csv')
print(iris.head())
print(iris.shape)
```

```python
X = iris.drop(columns=["Species"])
y=iris['Species']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42) #Training with 70% data, Testing with 30% data
#print(X_train.shape)
#print(X_test.shape)
#print(y_train.shape)
#print(y_test.shape)
scaler=StandardScaler()  #Preprocessing Operations
scaler.fit(X_train)
X_train=scaler.transform(X_train)
X_test=scaler.transform(X_test)
#print(X_train)
#print(X_test)
classifier=KNeighborsClassifier(n_neighbors=5)
classifier.fit(X_train,y_train)
y_pred=classifier.predict(X_test) #Predicted output(y_pred) Original output(y_test)
print("Classification Report: ")
print(classification_report(y_test,y_pred))
print("Accuracy: ")
print(accuracy_score(y_test,y_pred))
print("Confusion Matrix: ")
print(confusion_matrix(y_test,y_pred))
```
